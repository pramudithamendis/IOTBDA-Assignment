```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1. Load Data
# -----------------------------
# Replace with your file path
file_path = "output_with_movement_within_sleep_session.csv"

df = pd.read_csv(file_path)

# -----------------------------
# 2. Data Cleaning
# -----------------------------
# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Handle missing values (if any)
df = df.dropna()

# Convert movement column to numeric
df["movement_flag"] = df["movement"].map({"Y": 1, "N": 0})

# -----------------------------
# 3. Select Relevant Columns
# -----------------------------
numeric_cols = [
    "light_intensity",
    "temperature",
    "humidity",
    "sound_level",
    # "distance",
    # "movement_flag"
]

data = df[numeric_cols]

# -----------------------------
# 4. Correlation Matrix
# -----------------------------
corr_matrix = data.corr(method="pearson")

print("\nCorrelation Matrix:\n")
print(corr_matrix)

# -----------------------------
# 5. Visualization (Heatmap)
# -----------------------------
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True)
plt.title("Sensor Correlation Heatmap")
plt.show()

# -----------------------------
# 7. Optional: Save Results
# -----------------------------
# corr_matrix.to_csv("correlation_output.csv")

# print("\nCorrelation matrix saved as correlation_output.csv")
```

```python
# Convert movement to numeric
df["movement_flag"] = df["movement"].map({"Y": 1, "N": 0})

# Drop missing values if any
df = df.dropna()

# -----------------------------
# Features to analyze
# -----------------------------
features = [
    "light_intensity",
    "temperature",
    "humidity",
    "sound_level",
    "distance"
]

# -----------------------------
# Create box plots
# -----------------------------
for col in features:
    plt.figure(figsize=(6, 4))
    
    sns.boxplot(x="movement_flag", y=col, data=df)
    
    plt.title(f"{col} vs Movement")
    plt.xlabel("Movement (0 = No, 1 = Yes)")
    plt.ylabel(col)
    
    plt.show()
```
